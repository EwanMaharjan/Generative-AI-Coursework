```python
# Import libraries
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
import matplotlib.pyplot as plt
```

```python
#checking for cuda
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

```python
#pretrained model - stable diffusion
model_id = "runwayml/stable-diffusion-v1-5"

pipeline = StableDiffusionPipeline.from_pretrained(model_id)
pipeline = pipeline.to(device)

#text prompts to generate image
prompt = [
    "A munchkin cat in a forest",
    "A persian cat in a kitchen",
    "A siamese cat on a sofa"
]

guidance_scale = 7.5
steps = 2
num_images = 1

#image generation
result = pipeline(
prompt=prompt,
num_images_per_prompt=num_images,
guidance_scale=guidance_scale,
num_inference_steps = steps
)
image = result.images[0]
```

```python
#shows the image
fig, axes = plt.subplots(1, len(result.images), figsize=(5 * len(result.images), 5))

for i, (ax, image) in enumerate(zip(axes, result.images)):
    ax.imshow(image)
    ax.axis("off")
    image.save(f"generated_image_{i}.png")

plt.tight_layout()
plt.show()
```
