```python
#libraries
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
import matplotlib.pyplot as plt
```

```python
#checking for cuda
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

```python
#pretrained model - stable diffusion
model_id = "runwayml/stable-diffusion-v1-5"

pipeline = StableDiffusionPipeline.from_pretrained(model_id)
pipeline = pipeline.to(device)

#text prompts to generate image
prompt = [
    "A tabby cat on the street",
    "A tiger cat in a cave",
    "A siamese cat on a boat",
    "A egyptian cat in a desert"
]

guidance_scale = 7.5
steps = 50
num_images = 1

#image generation
result = pipeline(
prompt=prompt,
num_images_per_prompt=num_images,
guidance_scale=guidance_scale,
num_inference_steps = steps
)

image = result.images[0]
```

```python
#shows images side by side
n = len(result.images)
plt.figure(figsize=(5 * n, 5))

for i, (image, text) in enumerate(zip(result.images, prompt)):
    plt.subplot(1, n, i + 1)
    plt.imshow(image)
    plt.title(text, fontsize=9)
    plt.axis("off")
    image.save(f"generated_image_{i}.png")

plt.tight_layout()
plt.show()
```

```python
import open_clip

#clip score
model, _, preprocess = open_clip.create_model_and_transforms(
    "ViT-B-32", pretrained="laion2b_s34b_b79k", device=device
)
tokenizer = open_clip.get_tokenizer("ViT-B-32")


def clipscore_openclip(references, predictions):
    """Compute cosine similarity between text and image pairs."""
    assert len(references) == len(predictions), (
        "Text and image lists must match in length."
    )
    scores = []
    with torch.no_grad():
        for txt, img_path in zip(references, predictions):
            # Encode text
            text_tok = tokenizer([txt]).to(device)
            text_emb = model.encode_text(text_tok)
            text_emb /= text_emb.norm(dim=-1, keepdim=True)

            #encode
            image = Image.open(img_path).convert("RGB")
            image_tensor = preprocess(image).unsqueeze(0).to(device)
            img_emb = model.encode_image(image_tensor)
            img_emb /= img_emb.norm(dim=-1, keepdim=True)

            #cosine similarity
            cos_sim = (text_emb @ img_emb.T).item()
            scores.append((cos_sim + 1.0) / 2.0)

    return {"scores": scores, "average_score": float(np.mean(scores))}


references = [
    "A munchkin cat in a forest",
    "A bengal cat in a flowerfield",
    "A sphynx cat in a graveyard",
]
predictions = [
    "generated_images/generated_image.png",
    "generated_images/generated_image_7.png",
    "generated_images/generated_image_5.png",
]

clip_res = clipscore_openclip(references, predictions)
print("CLIP Similarity Results:", clip_res)
```
